# 🎚️ Stem Separation (Demucs) 
Run **Demucs** (or Spleeter) to separate audio into stems (vocals, drums, bass, other). Then preview and visualize waveforms.

**Requirements:**
- `pip install demucs librosa soundfile matplotlib`
- FFmpeg installed on your system for best results.

**Inputs:** set `INPUT_AUDIO`.
**Outputs:** stems saved under `outputs/stems/<model>/<track>/`.

```
# !pip install demucs librosa soundfile matplotlib
```

```
import os, subprocess, sys
from pathlib import Path
import librosa, soundfile as sf
import matplotlib.pyplot as plt
import numpy as np

INPUT_AUDIO = 'path/to/your/audio.wav'  # <-- change me
OUT_ROOT = Path('outputs/stems')
MODEL = 'htdemucs'  # common high-quality model

OUT_ROOT.mkdir(parents=True, exist_ok=True)
cmd = [sys.executable, '-m', 'demucs', '-n', MODEL, '-o', str(OUT_ROOT), INPUT_AUDIO]
print('Running:', ' '.join(cmd))
try:
    subprocess.run(cmd, check=True)
except Exception as e:
    print('Demucs failed:', e)
    print('Make sure demucs is installed and INPUT_AUDIO exists.')
```

## 🔎 Locate Output Stems

```
track_name = Path(INPUT_AUDIO).stem
# Demucs creates: outputs/stems/{MODEL}/{TRACK_NAME}/[drums.wav,bass.wav,other.wav,vocals.wav]
stem_dir = OUT_ROOT / MODEL / track_name
print('Stem dir:', stem_dir)
list(stem_dir.glob('*.wav')) if stem_dir.exists() else []
```

## 📈 Visualize Waveforms

```
def plot_wave(path):
    y, sr = librosa.load(path, sr=None)
    plt.figure(figsize=(12,2))
    plt.plot(y)
    plt.title(Path(path).name)
    plt.tight_layout()
    plt.show()

if stem_dir.exists():
    for wav in stem_dir.glob('*.wav'):
        plot_wave(str(wav))
else:
    print('No stems found. Check earlier cell output.')
```
